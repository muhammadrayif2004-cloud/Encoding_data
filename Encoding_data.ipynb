```python
df_std['Family_background'].unique()
```

```python

encoder_family = OrdinalEncoder(
    categories=[['Lower Middle class', 'Middle class', 'Rich class']]
)
df_std["Family_background"] = encoder_family.fit_transform(df_std[["Family_background"]])
```

```python
df_std.head(20)
```

```python
encoder = OneHotEncoder()
encoded_data = encoder.fit_transform(df_std[["Gender"]])

encoded_df_std = pd.DataFrame(
    encoded_data.toarray(),
    columns=encoder.get_feature_names_out(["Gender"]),
    index = df_std.index
    )

df_std = pd.concat([df_std, encoded_df_std], axis=1).drop(columns=["Gender"])
```
